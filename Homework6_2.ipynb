##Task 2(20 points):
We discussed how we can formulate RL problems as an MDP. Describe any
real-world application that can be formulated as an MDP. Describe the state space, action
space, transition model, and rewards for that problem. You do not need to be precise in the
description of the transition model and reward (no formula is needed). Qualitative description
is enough.

## Answer
A real-world application that can be formulated as a Markov Decision Process (MDP) is **autonomous driving**. In this problem, a self-driving car must navigate an environment while optimizing safety, efficiency, and adherence to traffic laws. The **state space** includes factors such as the car’s position (e.g., at an intersection or highway lane), velocity (e.g., moving at 60 mph or stopping at a red light), locations and speeds of nearby vehicles (e.g., a car merging from the right or a slow truck ahead), traffic signals (e.g., green, yellow, or red), and road conditions (e.g., wet or dry pavement). The **action space** consists of possible driving maneuvers like accelerating to merge onto a highway, braking to stop at a red light, changing lanes to overtake a slower vehicle, or making a left turn at an intersection.

The **transition model** captures how the car’s actions alter the state. For example, if the car decides to accelerate, its speed increases, and it moves forward faster; if another vehicle unexpectedly brakes ahead, the self-driving car may need to slow down to avoid a collision. External factors like pedestrians crossing or sudden weather changes also affect state transitions. The **reward function** incentivizes safe and efficient driving by assigning positive rewards for maintaining a steady speed (e.g., driving at 55 mph on a highway), obeying traffic rules (e.g., stopping at a red light), and reaching the destination efficiently. Conversely, penalties apply for unsafe behaviors such as sudden braking (e.g., causing discomfort for passengers), running a red light (e.g., risking an accident), or colliding with another vehicle (e.g., receiving a large negative reward). By formulating autonomous driving as an MDP, reinforcement learning can be used to train self-driving systems that learn optimal driving strategies in complex real-world scenarios.
